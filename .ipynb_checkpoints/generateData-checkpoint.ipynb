```python
import numpy as np
import pandas as pd
import os
```

```python
NUM_ROWS = 5000
OUTPUT_DIR = "data"
OUTPUT_FILE = "pricing_data.csv"
os.makedirs(OUTPUT_DIR, exist_ok=True)
```

```python
#multipliers
df = pd.DataFrame({
    'duration_hours':      np.random.randint(1, 6, size=NUM_ROWS),
    'lead_time_hours':     np.random.uniform(0.5, 168, size=NUM_ROWS),
    'zone_center':         np.random.choice([0, 1], size=NUM_ROWS),
    'severity_score':      np.random.randint(0, 4, size=NUM_ROWS),
    'special_reqs_count':  np.random.randint(0, 3, size=NUM_ROWS),
    'demand_supply_ratio': np.random.uniform(0.5, 3.0, size=NUM_ROWS)
})
```

```python
services = ["General House Cleaning",
      "Home Organizing",
      "Deep Cleaning",
      "Aircond Cleaning",
      "Carpet Cleaning",
      "Post-Renovation Cleaning",
      "Sofa or Mattress Cleaning",  
     "Plumbing Services",
      "Air Conditioner Repair",
      "Electrical Repair",
      "Washing Machine Repair",
      "Refrigerator Repair",
      "Door & Lock Repair",
      "Ceiling Repair", 
     "Furniture Assembly",
      "Mounting",
      "Painting & Touch-up Work",
      "Curtain or Blind Installation",
      "Minor Welding Jobs",
      "Kitchen Remodeling",
      "Tiling & Flooring",
      "Electrical Safety Check",
      "Gas Leak Detection",
      "Fire Extinguisher Servicing", 
     "House Moving",
      "Large Item Delivery",
      "Small Item Delivery", 
     "Lawn Mowing",
      "Gardening",
      "Tree Cutting",
      "Roof or Gutter Cleaning"
    ]

#base rate per hour or per unit
base_rates = {
    'general_house_cleaning': 13,
    'home_organizing': 16,
    'deep_cleaning': 22,
    'aircond_cleaning': 36,  
    'carpet_cleaning': 24,
    'post_renovation_cleaning': 48,
    'sofa_or_mattress_cleaning': 24,
    'plumbing_services': 26,
    'air_conditioner_repair': 36,
    'electrical_repair': 28,
    'washing_machine_repair': 34,
    'refrigerator_repair': 36,
    'door_lock_repair': 22,
    'ceiling_repair': 28,
    'furniture_assembly': 20,
    'mounting': 20,
    'painting_touch_up_work': 25,
    'curtain_or_blind_installation': 20,
    'minor_welding_jobs': 32,
    'kitchen_remodeling': 72,
    'tiling_flooring': 42,
    'electrical_safety_check': 32,
    'gas_leak_detection': 36,
    'fire_extinguisher_servicing': 34,
    'house_moving': 36,
    'large_item_delivery': 30,
    'small_item_delivery': 20,
    'lawn_mowing': 26,
    'gardening': 22,
    'tree_cutting': 38,
    'roof_or_gutter_cleaning': 44
}

service_choices = np.random.choice(services, size=NUM_ROWS)
df['service'] = service_choices
dummies = pd.get_dummies(df['service'], prefix='service')

# Replace all spaces to underscore and lowercase all letters
dummies.columns = (
    dummies.columns
           .str.lower()
           .str.replace(r'[^0-9a-z_]+','_', regex=True)
           .str.rstrip('_')
)

# One-hot encoding
df = pd.concat([df, dummies], axis=1)

# Compute base rate
def get_base_rate(row):
    for col in dummies.columns:
        if row[col] == 1:
            key = col.replace('service_', '')
            return base_rates.get(key, 0)
    return 0

df['base_rate'] = df.apply(get_base_rate, axis=1)

def duration_price(base_rate, duration):
    if duration <= 1:
        return base_rate
    elif duration == 2:
        return base_rate + (base_rate * 0.6)
    elif duration == 3:
        return base_rate + (base_rate * 0.6) + (base_rate * 0.5)
    else:
        return base_rate + (base_rate * 0.6) + (base_rate * 0.5) + (duration - 3) * (base_rate * 0.4)
# Price calculation
def final_price(row):
    base_duration_price = duration_price(row['base_rate'], row['duration_hours'])

    price = base_duration_price

    # Urgency multiplier
    urgency_multiplier = 1.0
    if row['lead_time_hours'] <= 24:
        urgency_multiplier = 1.3
    elif row['lead_time_hours'] <= 72:
        urgency_multiplier = 1.1
    price *= urgency_multiplier

    # Zone multiplier
    price *= 1.05 if row['zone_center'] == 1 else 1.0

    # Additive special request cost
    price += 1.5 * row['special_reqs_count']

    # Severity multiplier
    severity_multiplier = 1 + (row['severity_score'] * 0.1)
    price *= severity_multiplier

    # Surge pricing with cap
    surge_multiplier = min(max(row['demand_supply_ratio'], 0.8), 1.3)
    price *= surge_multiplier

    # Avoid having a lower price than the base rate
    min_price = base_duration_price
    max_price = base_duration_price * 1.8
    
    price = max(min_price, price)
    price = min(max_price, price)

    return round(price, 2)

df['price'] = df.apply(final_price, axis=1)

```

```python
# Save and preview
output_path = os.path.join(OUTPUT_DIR, OUTPUT_FILE)
df.to_csv(output_path, index=False)
print(f"Pricing data saved to: {output_path}")
print(df[['service', 'base_rate', 'price']].head())
```
